The bot should remember previously asked questions and answers and use them as context when generating new answers. [<span style="font-size: 0.6em">(go back to top)</span>](#top)

```python
# Specify API configuration
client = AzureOpenAI(
    api_key = os.getenv("OPENAPI_KEY"),
    api_version = '2025-02-01-preview',
    azure_endpoint = 'https://hkust.azure-api.net'
)
model_name = "gpt-5-mini"
# Initial system message
messages = [
    {"role": "system", "content": "You are an intelligent assistant."}
]

Qnum=1
# Question and answer loop
question = ""
while question != "Q":
    question = input("\nPlease input your question (input Q to quit): ")

    if question != "Q":
      messages.append({"role":"user", "content":question})
      response = client.chat.completions.create(
          model = model_name,
          messages = messages)
      print(f"Q{Qnum}: {question}")
      print(f"A{Qnum}: {response.choices[0].message.content}\n")
      Qnum+=1
    else:
        print("\n> User: ", question)
        print("> Model: Glad to help. Bye-bye")
```

#### **2. Basic Prompt Templates in LangChain** [<span style="font-size: 0.6em">(go back to top)</span>](#top)
<a href="https://reference.langchain.com/python/langchain/">LangChain</a> is an open-source orchestration framework that provides functions and components to connect LLMs with various data sources, enabling the creation of complex, multi-step workflows.


- **2.1 - LangChain's `PromptTemplate`** defines a resuable string template with an **explicit input schema**, allowing prompts to be filled dynamically at runtime, either **partially or fully**. This supports **standardized and reusable** prompt patterns across an application without repeatedly supplying identical input values.

```python
from langchain_core.prompts import PromptTemplate

# Define a prompt pattern
template_2a = PromptTemplate.from_template(
    "Suggest {num_cases} use cases of {technology} in {domain}."    
)

# Render the prompt with specific values
prompt_2a = template_2a.format(
    num_cases=3,
    technology="neural networks",
    domain="biology"
)
print(prompt_2a)
```

```python
# Get prompt variables
print(template_2a.input_variables) 

# Partial applications to "freeze" some variable
base_template_2b = template_2a.partial(
    num_cases=3,
    domain='marketing', 
)

topics = ['supervised learning', 'unsupervised learning', 'deep learning']
for t in topics:
    prompt_2b = base_template_2b.format(technology = t)
    print(prompt_2b)
```

- **2.2 - LangChain's `ChatPromptTemplate`** is designed for modern LLMs that use chat-style APIs with distinct roles such as *system*, *human*, and *ai*. It enables the definition of structured, multi-message templates for reliably constructing conversational inputs to chat-based models. ***Note***: Role names differ in OpenAI and LangChain packages. OpenAI uses `system`, `user`, `assistant`, while LangChain commonly uses `system`, `human`, `ai` (and some APIs also accept user/assistant). 

```python
from langchain_core.prompts import ChatPromptTemplate

template_2c = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a {role}. Always respond in {style} style."),
        ("human", "What is pizza?"),
        ("ai", "Pizza is a round(ish) flatbread covered in sauce, melted cheese, and questionable toppings, "
                "scientifically engineered to make you say “just one more slice” four times in a row."),
        ("human", "{followup_question}"),
    ]
)

prompt_2c = template_2c.format_messages(
    role="comedian",
    style="humorous and funny",
    followup_question="What's the most popular topping?",
)
#print("---", messages_2c, "\n")

for msg in prompt_2c:
    print("[{0}] {1}".format(msg.type.upper(), msg.content))
    #print(f"[{msg.type.upper()}] {msg.content}")
```

- **2.3  LangChain's `FewShotPromptTemplate`** makes it easy to manage examples separately from the template structure in **few-shot prompting**. 

```python
from langchain_core.prompts import FewShotPromptTemplate

examples = [
    {"review": "This product is amazing! Best purchase ever.", "sentiment": "Positive"},
    {"review": "Terrible quality. Broke after one day.", "sentiment": "Negative"},
    {"review": "It's okay. Nothing special but gets the job done.", "sentiment": "Neutral"},
    {"review": "Absolutely love it! Will buy again!", "sentiment": "Positive"},
    {"review": "Waste of money. Very disappointed.", "sentiment": "Negative"},
]

# Defind a template for formatting EACH example
example_template_2d = PromptTemplate(
    template="Review: {review}\nSentiment: {sentiment}",
    input_variables=["review", "sentiment"],
)

# Define a template for fewshot prompting
fewshot_template = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_template_2d,
    prefix="Classify the sentiment of the following product review as Positive, Negative, or Neutral.\n\n"
           "Here are some examples:", # general description before the examples
    suffix="Review: {input_review}\nSentiment:", # New input after the examples
    input_variables=["input_review"],
    example_separator="\n\n",
)

prompt_2d = fewshot_template.format(
    input_review="The color is nice but the zipper is flimsy."
)

print(prompt_2d)
```

#### **4. Prompt Chaining with LangChain** [<span style="font-size: 0.6em">(go back to top)</span>](#top)
**Pipelines** in LangChain are built from independent, reusable components called `Runnable`, which allow wiring multiple prompt + LLM steps together so that the output of one becomes the input to the other, i.e., **chaining** the prompts. 
- `Runnable` is the generic interface in LangChain for anything that can take input and produce output, such as an LLM, a prompt, a chain, and a tool.
 - `RunnableLambda` wraps a simple Python function as a runnable step in a LangChain pipeline
 - `RunnableMap` or `RunnableParallel` runs multiple branches in parallet on the same input, returning a dict of results.
 - `RunnableSequence` chains mutiple runnables in a sequential order, feeding the output of each into the next (can be constructed using the `|` operator)

 More details can be found https://reference.langchain.com/python/langchain_core/runnables/.
 


```python
from langchain_openai import AzureChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableSequence


# 1. LLM configured for Azure OpenAI
llm = AzureChatOpenAI(
    azure_endpoint=os.getenv("AZURE_OPENAI_ENDPOINT"),
    api_key=os.getenv("AZURE_OPENAI_API_KEY"),
    api_version=os.getenv("AZURE_OPENAI_API_VERSION"),

    model="gpt-4o"
)
# 2. First prompt: summarize text
summary_prompt = PromptTemplate.from_template(
    "Summarize the following text in 3–4 sentences:\n\n{text}"
)

# 3. Second prompt: turn the summary into bullet-point action items
actions_prompt = PromptTemplate.from_template(
    "Based on this summary, extract 3–5 clear action items as bullet points:\n\n{summary}"
)

# 4. Wrap LLM calls (prompt → LLM)
summary_step = summary_prompt | llm
actions_step = actions_prompt | llm

# 5. Compose them into a single chain using RunnableSequence
full_chain = RunnableSequence(summary_step, actions_step)

# 6. Invoke the chain with input text
input_text = """
Our company has grown rapidly over the past three years, expanding from a small startup 
of 15 people to a global team of more than 400 employees distributed across 10 countries. 
While this growth has been exciting, it has also introduced several operational challenges.

First, communication has become fragmented, with different teams using different tools, 
time zones making synchronous meetings difficult, and important decisions often buried 
in long message threads. Second, onboarding new employees has become inconsistent; 
some teams have thorough documentation and clear training plans, while others rely 
on ad-hoc explanations and tribal knowledge.

Third, we lack a unified system for setting and tracking goals. Teams often define 
their own priorities without clear alignment to company-wide objectives, which leads 
to duplicated efforts and confusion about what success looks like. Finally, many 
employees report feeling uncertain about career growth opportunities and expectations 
for promotion, which impacts motivation and retention.

Leadership wants to create a more scalable, transparent, and coherent way of working. 
This includes improving documentation, standardizing onboarding, clarifying company-wide 
goals, and offering clearer career paths, while still preserving the flexibility and 
autonomy that made the company successful in its early days.
"""
result = full_chain.invoke({"text": input_text})

print("Proposed action items:\n" + result.content)
```

- **2.3 Similarity between token embeddings**
  - Tokens with similar meanings or usage contexts tend to have embedding vectors that are close to each other in the embedding space.

**This is to see how similar the words are**

```python
import torch

# Assume token1 and token2 are single tokens (i.e., each becomes EXACTLY ONE vocab item).
# If a "word" gets split into subwords (e.g., ["app", "##le"]), this function will
# produce multiple embedding vectors and the cosine similarity call will return multiple values depending on shapes.
def get_similarity(tokenizer, model, token1, token2):
    id1 = tokenizer.encode(token1, add_special_tokens=False)
    id2 = tokenizer.encode(token2, add_special_tokens=False)
    
    embedding_layer = model.get_input_embeddings()  # nn.Embedding
    
    embedding1 = embedding_layer(torch.tensor(id1)) # shape: [1, 768]
    embedding2 = embedding_layer(torch.tensor(id2)) # shape: [1, 768]
    #print("---", embedding1.shape)
    
    score = torch.nn.functional.cosine_similarity(embedding1, embedding2, dim=1) # compare along the embedding dimension (768)

    return score.detach().numpy() # convert tensor([score]) -> numpy float

# Similar tokens should have higher cosine similarity in embedding space
print(get_similarity(bert_tokenizer, bert_model, "dog", "cat")) # High similarity 
print(get_similarity(bert_tokenizer, bert_model, "dog", "pen")) # Low similarity 

# BUT: these are static token embeddings in the *input embedding table*,
# not contextual embeddings from a sentence.
print(get_similarity(bert_tokenizer, bert_model, "apple", "computer"))  
print(get_similarity(bert_tokenizer, bert_model, "lemon", "apple"))
```

- **2.4 Contextual embeddings**
  - In practice, language models produce contextual embeddings: the vector for the same token can change depending on the surrounding words, capturing different meanings in different contexts.

```python
import torch

s1 = "I went to the bank to deposit money."
s2 = "The river bank was muddy."

# 1. Prepare inputs (This returns the required MAPPING)
inputs1 = bert_tokenizer(s1, return_tensors="pt")
inputs2 = bert_tokenizer(s2, return_tensors="pt")
#print(inputs1)
#print(inputs2)

# 2. Find indices (need to convert the IDs back to tokens to find 'bank')
tokens1 = bert_tokenizer.convert_ids_to_tokens(inputs1.input_ids[0])
tokens2 = bert_tokenizer.convert_ids_to_tokens(inputs2.input_ids[0])

word1 = "bank" # Pick up word from s1
word2 = "bank" # Pick up word from s2
idx1 = tokens1.index(word1)
idx2 = tokens2.index(word2)

# 3. Pass the specific tensors of the inputs to the model to get the output
with torch.no_grad():    
    outputs1 = bert_model(input_ids=inputs1["input_ids"], attention_mask=inputs1["attention_mask"]) 
    outputs2 = bert_model(input_ids=inputs2["input_ids"], attention_mask=inputs2["attention_mask"])

# 4. Extract embedding vectors from the model's last hidden layer [batch, seq_len, embedding_dim] 
vec1 = outputs1.last_hidden_state[0, idx1, :] 
vec2 = outputs2.last_hidden_state[0, idx2, :]

## print out the token embeddings, first a new values
print(f"Token {idx1} embeddings in s1: {vec1[0:20]}") 
print(f"Token {idx2} embeddings in s2: {vec2[0:20]}")

# 5. Similarity
similarity = torch.nn.functional.cosine_similarity(vec1.unsqueeze(0), vec2.unsqueeze(0))
print(f"Similarity between '{word1}' in s1 and '{word2}' in s2: {similarity.item():.4f}")
```

#### **3. Sentence Embeddings and Similarity** [<span style="font-size: 60%">(go back to top)</span>](#top)

- `SentenceTransformers` is a Python framework for state-of-the-art sentence, text and image embeddings. It provides easy-to-use pretrained models that map variable-length text into fixed-length vectors, enabling tasks like semantic similarity, clustering, and semantic search with cosine similarity. <br/>Details can be found at: https://www.sbert.net/

```python
!pip3 install sentence_transformers
```

```python
from sentence_transformers import SentenceTransformer

sentence_model = SentenceTransformer('all-MiniLM-L6-v2') # balanced bwteen speed and quality

sentences = [
    "Transformers are powerful models.",
    "Cats sit on mats.",
    "Transformers are neural networks for sequences.",
    "I enjoy studying deep learning."
]

# Compute embeddings
embeddings = sentence_model.encode(sentences, convert_to_tensor=True)
#print("---", embeddings.shape)

# Compute cosine similarity between two sentences
print("--- Embeddings for s[0], ", embeddings[0:1, :].shape)
sim_0_1 = torch.nn.functional.cosine_similarity(embeddings[0:1, :], embeddings[1:2, :], dim=1) # between sentence 0 and 1
print(f"Similarity (s0 vs s1): {sim_0_1.item():.4f}\n")

# Normalize embeddings to unit length (makes cosine similarity a simple dot product) 
# norm_embeddings shape: (doc_n, embed_dim)
norm_embeddings = torch.nn.functional.normalize(embeddings, p=2, dim=1)

# Multiply the matrix by its own transpose
# shape: (doc_n, embed_dim) x (embed_dim, doc_n) -> (doc_n, doc_n)
sim_matrix = norm_embeddings @ norm_embeddings.T  # 
print("Similarity matrix:", sim_matrix.shape)

torch.set_printoptions(precision=4, sci_mode=False)
print(sim_matrix)
```

|---------|---------|--------------|
| 1 | Tokenizer Basics | Tokenizers convert raw text into **tokens** (words/subwords/bytes) and then into **token IDs**. Different tokenizers behave differently: **BERT WordPiece** may produce **`[UNK]`** for out-of-vocabulary characters; **GPT‑2 byte-level BPE** typically avoids OOV by operating at the byte level. |
| 2 | From Tokens to Embeddings | Language models store an **embedding table** that maps each token ID to a dense vector. The input embedding output has shape \([batch, seq_len, emb_dim]\). If you **add new tokens** to a tokenizer, you must also **resize the model’s token embeddings** to match the new vocabulary size. |
| &nbsp; | Static vs Contextual Embeddings | The **input embedding table** gives **static** embeddings (one vector per token, regardless of context). In contrast, **contextual embeddings** come from the model’s hidden states (e.g., `last_hidden_state`) and change depending on surrounding words. |
| 3 | Sentence Embeddings & Similarity | `SentenceTransformers` produces **fixed-size sentence vectors** suitable for semantic comparison. Cosine similarity can be computed pairwise or as a full **similarity matrix** efficiently by **normalizing embeddings** and using a dot product. |

#### **1. Generating Responses with a Language Model** [<span style="font-size: 0.6em">(go back to top)</span>](#top)

- **1.1** Download and initialize a pre-trained model

```python
DEVICE = torch.device("cpu")
MODEL_NAME = "distilgpt2"  # fallback: "sshleifer/tiny-gpt2"

tokenizer = GPT2Tokenizer.from_pretrained(MODEL_NAME)
base_model = GPT2LMHeadModel.from_pretrained(MODEL_NAME)

# GPT-2 has no pad token by default — reuse the end-of-text token
tokenizer.pad_token = tokenizer.eos_token
base_model.config.pad_token_id = tokenizer.eos_token_id

print(f"Model : {MODEL_NAME}")
print(f"Parameters : {base_model.num_parameters():,}")
#print(model)
#print(model.config)
```

- **1.2** Response generation: base model

```python
#@torch.no_grad()
def generate_text(model, prompt, max_new_tokens=80, temperature=0.8, top_p=0.95):
    model.eval()
    inputs = tokenizer(prompt, return_tensors="pt").to(DEVICE)

    # Always return teh full sequence including input tokens
    out = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=True,
        temperature=temperature,
        top_p=top_p,
        pad_token_id=tokenizer.pad_token_id,
        eos_token_id=tokenizer.eos_token_id,
    )

     # Only decode the NEW tokens, not include the prompt
    new_tokens = out[0][inputs["input_ids"].shape[-1]:] #
    return tokenizer.decode(new_tokens, skip_special_tokens=True)
```

- **2.1** Build a continued pre-training (CPT) dataset

```python
domain_texts = [
    "Class participation: this component of the assessment is based on in-class engagement and the quality of contributions to open discussion. Some pop-up quizzes will be conducted in class to evaluate students' understanding of the knowledge and concepts discussed in lectures.",
    "Hands-on labs: students are required to use Python and/or AI agent framework/platform to accomplish a specific LLM deployment or prototyping task.",
    "Group project: students will form a group to conduct research and present their findings on a specific LLM-related topic. Groups may also propose a topic of their own interest, subject to the instructor's approval.",
    "Late-submission policy: late submissions receive a 20% penalty per day unless an extension is approved.",
    "Office hours are for debugging, conceptual questions, and feedback on project direction.",
    "Grading breakdown: class participation accounts for 15%, hands-on labs account for 25%, the group project accounts for 20%, and the final exam accounts for 40%.",
    "Academic integrity: all submitted work must be original. Use of AI-generated content must be disclosed and properly attributed. Violations will be reported to the academic integrity office.",
    "Prerequisites: students should have basic Python programming experience and familiarity with machine learning concepts. No prior NLP experience is required.",
    "Recommended textbooks: there is no required textbook. Recommended readings are included in the lecture slides."
    "Group project: each project group must consist of 2 to 4 members. Groups must be formed on Canvas by the announced due date.",
    "Regrade requests: students may submit a regrade request within 7 days of receiving a graded assignment. Requests must include a written justification explaining why the grade should be reconsidered.",
] * 1
random.shuffle(domain_texts)

# Tokenization
input_ids =  tokenizer(domain_texts, add_special_tokens=False)["input_ids"]

# Flatten into one long list of token ids
flat = [tok_id for seq in input_ids for tok_id in seq]

# Truncate the token list to a multiple of BLOCK, so every training example has the same fixed length (seq_len = BLOCK)
BLOCK = 64
n = (len(flat) // BLOCK) * BLOCK
flat = flat[:n]
blocks = [flat[i:i + BLOCK] for i in range(0, n, BLOCK)]

# Build a dataset for LM training (labels same as input_ids for Causal Models)
ds_cpt = Dataset.from_dict({"input_ids": blocks, "labels": blocks})
ds_cpt.set_format(type="torch", columns=["input_ids", "labels"])

print("Num blocks:", len(ds_cpt))
print("Example shape:", ds_cpt[0]["input_ids"].shape, ds_cpt[0]["labels"].shape)

# Create a Dataset Loader to iterate batches during training
cpt_loader = DataLoader(ds_cpt, batch_size=1, shuffle=True)
```

 - **2.2** Continued pre-training (CPT)

```python
import copy

cpt_model = copy.deepcopy(base_model)

# Set the optimizer and loss function
cpt_optimizer = torch.optim.AdamW(cpt_model.parameters(), lr=5e-5)
cpt_model.config.loss_type = "ForCausalLMLoss"

# Set the number of epochs
# IMPORTANT, can change epochs
#CPT_EPOCHS = 5
CPT_EPOCHS = 10

_ = cpt_model.train() # set the model to training mode
for epoch in range(1, CPT_EPOCHS + 1):
    total_loss = 0.0
    for batch in cpt_loader:
        input_ids = batch["input_ids"].to(DEVICE)
        labels = batch["labels"].to(DEVICE)

        cpt_optimizer.zero_grad(set_to_none=True) # clear the gradients for each batch
        outputs = cpt_model(input_ids=input_ids, labels=labels)

        loss = outputs.loss
        loss.backward()
        cpt_optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(cpt_loader)
    print(f"  ✓ [CPT] Epoch {epoch}/{CPT_EPOCHS} complete — Avg Loss: {avg_loss:.4f}\n")
```

```python
# SFT Tokenization
MAX_LEN = 128 # input sequence length cap
EOS = tokenizer.eos_token
PAD_ID = tokenizer.pad_token_id

all_input_ids = []
all_attention_masks = []
all_labels = []

for ex in sft_examples:
    prompt = f"Instruction:{ex['instruction']}\nResponse:"
    answer = ex["response"].strip() + EOS # add an explicit end-of-sequence token
    full = prompt + answer

    # Tokenize prompt alone (to know where the response starts)
    prompt_ids = tokenizer(prompt, add_special_tokens=False)["input_ids"]

    # Tokenize the full text (prompt + answer)
    enc = tokenizer(full, add_special_tokens=False, truncation=True, max_length=MAX_LEN)
    input_ids = enc["input_ids"]

    # Pad to MAX_LEN (like BLOCK in CPT, every example has the same fixed length)
    # -100 values in labels tell PyTorch's loss function to skip those positions entirely
    pad_len = MAX_LEN - len(input_ids)
    attention_mask = [1] * len(input_ids) + [0] * pad_len

    labels = list(input_ids)
    labels += [-100] * pad_len                        # padding positions also set to -100, to be ingored in loss computation
    input_ids = input_ids + [PAD_ID] * pad_len

    all_input_ids.append(input_ids)
    all_attention_masks.append(attention_mask)
    all_labels.append(labels)

# Build a dataset for SFT training (labels masked on prompt portion)
ds_sft = Dataset.from_dict({
    "input_ids": all_input_ids,
    "attention_mask": all_attention_masks,
    "labels": all_labels,
})
ds_sft.set_format(type="torch", columns=["input_ids", "attention_mask", "labels"])

print("Num SFT examples:", len(ds_sft))
print("Example input_ids length:", len(ds_sft[0]["input_ids"]))
print("Example labels:", ds_sft[0]["labels"].tolist())
print("Example Attention Mask:", ds_sft[0]["attention_mask"].tolist())


# Create a DataLoader to iterate batches during training
sft_loader = DataLoader(ds_sft, batch_size=5, shuffle=True)
```

- **3.2** SPT traning (continues from CPT weights)

```python
sft_model = copy.deepcopy(cpt_model)

# Set the optimizer and loss function
sft_optimizer = torch.optim.AdamW(sft_model.parameters(), lr=5e-5)
sft_model.config.loss_type = "ForCausalLMLoss"

# Set the number of epochs
SFT_EPOCHS = 3

_ = sft_model.train()
for epoch in range(1, SFT_EPOCHS + 1):
    total_loss = 0.0
    for batch in sft_loader:
        input_ids = batch["input_ids"].to(DEVICE)
        labels = batch["labels"].to(DEVICE)
        masks = batch["attention_mask"].to(DEVICE)

        sft_optimizer.zero_grad(set_to_none=True) # clear the gradients for each batch
        outputs = sft_model(input_ids=input_ids, labels=labels, attention_mask=masks)

        loss = outputs.loss
        loss.backward()
        sft_optimizer.step()

        total_loss += loss.item()
        del outputs, loss # free the computation resources immediately

    avg_loss = total_loss / len(sft_loader)
    print(f"  ✓ [SFT] Epoch {epoch}/{SFT_EPOCHS} complete — Avg Loss: {avg_loss:.4f}\n")
```

  In a case like this, I would choose context engineering. If the domain text changes frequently, fine-tuning is too slow because you'd need to retrain, evaluate, and redeploy every time something changes. With context engineering you just update the text and it's immediately reflected in the next inference call. Fine-tuning also risks serving stale answers since the knowledge gets frozen into the weights at training time.

- **(d) Describe one possible hybrid strategy, explain when and why you would use it.**
  
  In my oppinion, a good hybrid strategy would be to fine-tune the model first (CPT + SFT) so it learns the domain knowledge and response format, and then also provide the relevant context in the prompt at inference time. This way the model already has the domain knowledge baked into its weights from training, but the prompt context can cover any new or updated information that wasn't in the training data. This makes sense when the domain knowledge is mostly stable but gets small updates occasionally, since you don't need to retrain the whole model every time something changes.

- **3.5** Evaluation with the metric of perplexity metric
> **Note**: Perplexity meansures how *surprised* the model is by the tokens (i.e., at each position, what probability does the model assign to the correct next token).
$\displaystyle \text{Perplexity} = \exp\left(-\frac{1}{N}\sum_{i=1}^{N} \log P(w_i \mid w_1, w_2, \dots, w_{i-1})\right)$ <br/>
Since cross-entropy loss is already the average negative log-likelihood: $\displaystyle\text{Loss} = -\frac{1}{N}\sum_{i=1}^{N} \log P(w_i \mid w_1, w_2, \dots, w_{i-1})$ <br/>
Thus, perplexity simplifies to: $\displaystyle \text{Perplexity} = e^{\text{Loss}}$
Lower perplexity is better.

```python
# Held-out text the model has NOT been trained on
# Cell 17: Perplexity evaluation — use instruction-formatted held-out examples
held_out_texts = [
    "Instruction:What is the attendance policy?\nResponse:##Attendance policy##: attendance is mandatory. Students who miss more than three classes without valid justification will receive a grade penalty of one letter grade.",
    "Instruction:How will course announcements be communicated?\nResponse:##Announcements##: all course announcements will be posted on the course LMS. Students are expected to check the LMS at least once per day during the semester.",
    "Instruction:What should I visit in Tokyo?\nResponse:In Tokyo, you should visit the Senso-ji temple in Asakusa, explore the Shibuya crossing, and try fresh sushi at Tsukiji outer market. The Meiji Shrine and Akihabara district are also worth a visit.",
]

# Compute perlexity on the response tokens.
# The loss is masked to ignore all tokens before (and including) the reponse marker
@torch.no_grad()
def compute_perplexity(model, texts, response_marker="\nResponse:"):
    model.eval()
    total_loss = 0.0
    total_tokens = 0

    for text in texts:
        # Tokenize the full text
        enc = tokenizer(text, return_tensors="pt").to(DEVICE)
        input_ids = enc["input_ids"]  # (1, seq_len)

        # Find where the response starts in token space
        # Tokenize only the instruction portion (up to and including the marker)
        marker_pos = text.find(response_marker)
        instruction_part = text[: marker_pos + len(response_marker)]
        instruction_ids = tokenizer(instruction_part, return_tensors="pt")["input_ids"]
        instruction_len = instruction_ids.shape[1]

        # Build labels: -100 for instruction tokens, real ids for response tokens
        labels = input_ids.clone()
        labels[0, :instruction_len] = -100

        # Count only the response tokens (non -100) in the sample text
        num_response_tokens = (labels != -100).sum().item()

        # Forward pass — model internally ignores -100 positions in loss
        outputs = model(input_ids=input_ids, labels=labels)

        # outputs.loss is already averaged over non-ignored tokens
        total_loss += outputs.loss.item() * num_response_tokens
        total_tokens += num_response_tokens

    avg_loss = total_loss / total_tokens
    return torch.exp(torch.tensor(avg_loss)).item()

base_ppl = compute_perplexity(model_base, held_out_texts)
full_ppl = compute_perplexity(model_spt, held_out_texts)
lora_ppl = compute_perplexity(model_lora, held_out_texts)
quant_ppl = compute_perplexity(model_quantized, held_out_texts)

print(f"{'Model':20s} {'Perplexity':>12s}")
print(f"{'Base (no training)':20s} {base_ppl:>12.2f}")
print(f"{'Full SFT':20s} {full_ppl:>12.2f}")
print(f"{'LoRA SFT':20s} {lora_ppl:>12.2f}")
print(f"{'Quantized SFT':20s} {quant_ppl:>12.2f}")
```

```python
from langchain_openai import AzureChatOpenAI

# --- LLM client and database query retriever ---
llm = AzureChatOpenAI(
    azure_endpoint=os.getenv("AZURE_OPENAI_ENDPOINT"),
    api_key=os.getenv("AZURE_OPENAI_API_KEY"),
    api_version=os.getenv("AZURE_OPENAI_API_VERSION"),
    model=OPENAI_CHAT_MODEL
)

retriever = vectorstore.as_retriever(
    search_type="similarity", # another option: "mmr"
    search_kwargs={"k": 4},
)

print("LLM and Retriever are ready.")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# --- Prompt template ---
RAG_PROMPT_TEMPLATE = """\
You are a helpful assistant that answers questions based ONLY on the provided context.
If the context does not contain enough information to answer, say "I don't have enough
information in the provided documents to answer this question."

Context:
{context}

Question: {question}

Answer:"""

rag_prompt = ChatPromptTemplate.from_template(RAG_PROMPT_TEMPLATE)

# --- Chunk formatter ---
def format_docs(docs):
    formatted = []
    for i, doc in enumerate(docs):
        page = doc.metadata.get("page", "?")
        formatted.append(f"[Chunk {i+1}, Page {page}]\n{doc.page_content}")
    return "\n\n---\n\n".join(formatted)

# --- RAG Chain ---
# '|' operator works like a pipe passing the output of the left to the right as input
rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough(), # pass the question as it is
    }
    # Basically, this is the order for the execution
    | rag_prompt
    | llm
    | StrOutputParser() # simply get the clean string out of the message object.
)

print("RAG chain built.")

my_questions = [
    "What is the procedures for dealing with student academic dishonesty cases found in final examinations? Answer in bulletin form.",
    "How much is the tuition fee for non-local students?" # (unrelated to the PDF)",
]

for q in my_questions:
    print(f"\n{'='*60}")
    print(f"{q}")
    print(f"{'='*60}")
    ans = rag_chain.invoke(q)  # q -> retriever -> format_docs() -> rag_prompt -> llm -> output parser
    print(f"{ans}\n")

```

- **3.2 Multi-turn conversional RAG bot**

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage

CONVERSATIONAL_PROMPT = """\
You are a helpful assistant that answers questions based ONLY on the provided context.
If the context is insufficient, say so.

Context:
{context}
"""

conversational_prompt = ChatPromptTemplate.from_messages([
    ("system", CONVERSATIONAL_PROMPT),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{question}"),
])

chat_history = []

def conversational_rag(question: str) -> str:
    """Ask a question with conversation memory."""
    docs = retriever.invoke(question)
    context = format_docs(docs)

    chain = conversational_prompt | llm | StrOutputParser()

    answer = chain.invoke({
        "context": context,
        "chat_history": chat_history,
        "question": question,
    })

    chat_history.append(HumanMessage(content=question))
    chat_history.append(AIMessage(content=answer))
    return answer

print("Conversational RAG function ready.\n")

# --- Demo conversation ---
chat_history.clear()

q1 = "What is this document about?"
a1 = conversational_rag(q1)
print(f"- {q1}\n{a1}\n")

q2 = "Can you give more details on that?"
a2 = conversational_rag(q2)
print(f"- {q2}\n{a2}\n")

q3 = "Summarize the key points in three bullet points."
a3 = conversational_rag(q3)
print(f"- {q3}\n {a3}\n")
```

```python
# --- Chunking configurations to compare ---
chunking_configs = [
    {"chunk_size": 400, "chunk_overlap": 50,  "label": "Small (400/50)"},
    {"chunk_size": 800, "chunk_overlap": 200, "label": "Medium (800/200)"},
    {"chunk_size": 1200, "chunk_overlap": 300, "label": "Large (1200/300)"},
]

# Step 1: Create the 5 questions with ground truth
eval_data = [
    {
        "question": "What penalties can be imposed for student misconduct?",
        "ground_truth": "Penalties include fine, university community service, withdrawal of privileges, suspension, or expulsion from the University, depending on how serious it is."
    },
    {
        "question": "What should a student do if they are sick during an examination?",
        "ground_truth": "Students should report to the University Health Services or seek medical assistance as soon as possible. They must provide appropriate documentation."
    },
    {
        "question": "What is the procedure for requesting a variance from academic regulations?",
        "ground_truth": "Requests should be made to the student's major department with supporting documents, then submitted to the appropriate University authority. Appeals can be made within two weeks."
    },
    {
        "question": "What are the rules regarding intellectual property at HKUST?",
        "ground_truth": "The university has policies on Copyright, Patents, Software, and Trade and Service Marks. All students must agree to these policies as a condition to be part of the school."
    },
    {
        "question": "How are academic dishonesty cases handled during examinations?",
        "ground_truth": "People in charge inform the student, document the incident on a Report on Student Academic Misconduct, and forward it to the course department for follow-up after the exam is finished."
    },]
```

```python
# --- Step 2: Function to evaluate a single chunking configuration ---
def evaluate_chunking_config(chunk_size, chunk_overlap, label):
    """Evaluate RAG performance with given chunking parameters"""
    print(f"\nEvaluating: {label}")
    # Create splitter - From Section 2.1 (Chunking)
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        length_function=len,
        separators=["\n\n", "\n", ". ", " ", ""],)

    # Split documents - raw_documents from Section 1 (Load PDF)
    chunks = splitter.split_documents(raw_documents)
    print(f"  Created {len(chunks)} chunks")

    # Create vectorstore - embedding_model from Section 2.2 (Vector Database)
    vectorstore = Chroma.from_documents(
        documents=chunks,
        embedding=embedding_model,)

    # Create retriever - From Section 3.1 (RAG Chain)
    retriever = vectorstore.as_retriever(
        search_type="similarity",
        search_kwargs={"k": 4},)

    # Create RAG chain - format_docs, rag_prompt, llm from Section 3.1 (RAG Chain)
    rag_chain_temp = (
        {
            "context": retriever | format_docs,
            "question": RunnablePassthrough(),
        }
        | rag_prompt
        | llm
        | StrOutputParser())

    # Prepare evaluation samples
    samples = []
    for item in eval_data:
        question = item["question"]
        ground_truth = item["ground_truth"]

        # Get contexts and answer
        retrieved_docs = retriever.invoke(question)
        contexts = [doc.page_content for doc in retrieved_docs]
        answer = rag_chain_temp.invoke(question)

        # SingleTurnSample from Section 4 (Evaluation)
        samples.append(SingleTurnSample(
            user_input=question,
            response=answer,
            retrieved_contexts=contexts,
            reference=ground_truth,
        ))

    # Run evaluation - metrics and evaluator from Section 4 (Evaluation)
    eval_dataset = EvaluationDataset(samples=samples)

    result = evaluate(
        dataset=eval_dataset,
        metrics=[
            Faithfulness(llm=evaluator_llm),
            ContextPrecision(llm=evaluator_llm),
            ContextRecall(llm=evaluator_llm),
            AnswerSimilarity(embeddings=evaluator_embeddings),
            AnswerCorrectness(llm=evaluator_llm, embeddings=evaluator_embeddings),
        ],
        raise_exceptions=False,
    )

    df = result.to_pandas()
    return {
        "label": label,
        "num_chunks": len(chunks),
        "faithfulness": df['faithfulness'].values[0],
        "context_precision": df['context_precision'].values[0],
        "context_recall": df['context_recall'].values[0],
        "answer_similarity": df['answer_similarity'].values[0],
        "answer_correctness": df['answer_correctness'].values[0],
    }
```

```python
print(f"Evaluating {len(eval_data)} questions across {len(chunking_configs)} configurations...")

results = []
for config in chunking_configs:
    try:
        results.append(evaluate_chunking_config(
            config["chunk_size"],
            config["chunk_overlap"],
            config["label"]
        ))
    except Exception as e:
        print(f"  Error: {e}")
        continue

# --- Step 4: Create summary table
import pandas as pd
df_results = pd.DataFrame(results)
# Format to 4 decimals
for col in ['faithfulness', 'context_precision', 'context_recall', 'answer_similarity', 'answer_correctness']:
    df_results[col] = df_results[col].apply(lambda x: f"{x:.4f}")
print(df_results.to_string(index=False))
```

- **2.2 Define Tools**

Four tools are needed and built. Each will be decorated with LangChain's `@tool`, which automatically generates the JSON schema the LLM needs from the function signature and docstring.
 - `get_sheet_names`: discover which sheets exist in the workbook
 - `get_schema`: learn column letters, header names, data types, and sample values for a sheet
 - `get_cell_range`: peek at specific cells when the LLM needs more context
 - `validate_formula`: parse, compile, and evaluate the formula against real data using the `formulas` library
   - This tool is utlizing the `formulas` library to validate the formula syntax and calculate the result.


```python
import json
import shutil
from langchain_core.tools import tool
from datetime import date, datetime
from openpyxl.utils import get_column_letter
from formulas import ExcelModel

# -- A helper function (not a tool — just used internally) -----
def format_value(val):
    """Convert a cell value to a JSON-friendly string."""
    if val is None:
        return "None"
    if isinstance(val, datetime):
        return val.strftime("%Y-%m-%d")
    if isinstance(val, date):
        return val.isoformat()
    if hasattr(val, "item"):  # numpy scalar
        return str(val.item())
    return str(val)


# -- Tool 1 -----------------------------------------------------
@tool
def get_sheet_names() -> str:
    """Return all sheet names in the Excel workbook. Call this first to discover the workbook structure."""
    try:
        wb = openpyxl.load_workbook(FILE_PATH, read_only=True)
        names = wb.sheetnames
        wb.close()
        return json.dumps({"sheets": names})
    except FileNotFoundError:
        return json.dumps({"error": f"File '{FILE_PATH}' not found."})


# -- Tool 2 -----------------------------------------------------
@tool
def get_schema(sheet_name: str) -> str:
    """Return the column layout of a GIVEN sheet: column letters, header names, data types, and sample values. Use this to understand what data is available before writing a formula."""
    wb = openpyxl.load_workbook(FILE_PATH, data_only=True)
    try:
        ws = wb[sheet_name]
    except KeyError:
        wb.close()
        return json.dumps({"error": f"Sheet '{sheet_name}' not found. Use get_sheet_names first."})

    max_row = ws.max_row
    max_col = ws.max_column
    columns = []

    for col_idx in range(1, max_col + 1):
        header = ws.cell(row=1, column=col_idx).value
        col_letter = get_column_letter(col_idx)

        samples = []
        for row_idx in range(2, min(7, max_row + 1)): # a few samples
            samples.append(ws.cell(row=row_idx, column=col_idx).value)

        non_none = [s for s in samples if s is not None]
        if not non_none:
            dtype = "empty"
        elif all(isinstance(v, (int, float)) for v in non_none):
            dtype = "numeric"
        elif all(isinstance(v, (date, datetime)) for v in non_none):
            dtype = "date"
        else:
            dtype = "text"

        columns.append({
            "column_letter": col_letter,
            "header": str(header),
            "data_type": dtype,
            "sample_values": [format_value(v) for v in samples[:3]],
        })

    wb.close()
    return json.dumps({
        "sheet_name": sheet_name,
        "total_rows": max_row,
        "data_rows": max_row - 1,
        "total_columns": max_col,
        "columns": columns,
    })


# -- Tool 3 -----------------------------------------------------
@tool
def get_cell_range(sheet_name: str, cell_range: str) -> str:
    """Return the values in a specific cell or range (e.g. 'A1:C5', 'D1:D10', 'B2'). Use this to see actual data such as unique values in a column."""
    wb = openpyxl.load_workbook(FILE_PATH, data_only=True)
    try:
        ws = wb[sheet_name]
    except KeyError:
        wb.close()
        return json.dumps({"error": f"Sheet '{sheet_name}' not found."})

    try:
        selection = ws[cell_range]
    except Exception as e:
        wb.close()
        return json.dumps({"error": f"Invalid range '{cell_range}': {e}"})

    if hasattr(selection, "value"):
        result = [[format_value(selection.value)]]
    elif isinstance(selection[0], tuple):
        result = [[format_value(c.value) for c in row] for row in selection]
    else:
        result = [[format_value(c.value) for c in selection]]

    wb.close()
    return json.dumps({"range": cell_range, "values": result})


# -- Tool 4 -----------------------------------------------------
EXCEL_ERRORS = {"#NAME?", "#REF!", "#VALUE!", "#DIV/0!", "#NULL!", "#N/A", "#NUM!", "#SPILL!", "#CALC!"}

@tool
def validate_formula(sheet_name: str, formula: str, target_cell: str = "A1") -> str:
    """Validate an Excel formula by parsing and compiling it with a real formula engine, then evaluating it against the actual workbook data to produce a sample result. Always call this before returning a formula to the user."""
    tmp_path = FILE_PATH.replace(".xlsx", "_tmp_validate.xlsx")

    try:
        # Create a temp copy the original file
        shutil.copy2(FILE_PATH, tmp_path)
        wb = openpyxl.load_workbook(tmp_path)
        ws = wb[sheet_name]
        ws[target_cell] = formula
        wb.save(tmp_path)
        wb.close()

        # Calculate the formulas inside
        xl_model = ExcelModel().loads(tmp_path).finish()
        results = xl_model.calculate()

        # Locate the target cell in the results dictionary
        sample_value = None
        for key, val in results.items():
            key_str = str(key).upper()
            if sheet_name.upper() in key_str and target_cell.upper() in key_str:
                if hasattr(val, "ravel"):
                    flat = val.ravel()  # flatten a multi-dimensional array into 1-D array
                    sample_value = flat[0] if len(flat) > 0 else None
                else:
                    sample_value = val
                break # break the loop when found the target cell

        if sample_value is not None:
            formatted = format_value(sample_value)

            # Check if the evaluated result is an Excel error
            is_error = any(err in str(formatted).upper() for err in EXCEL_ERRORS)
            if is_error:
                return json.dumps({
                    "valid": False,
                    "sample_result": formatted,
                    "evaluated_at": f"{sheet_name}!{target_cell}",
                    "message": f"Formula compiled but evaluated to an Excel error: {formatted}. "
                               f"This usually means a function name is misspelled, "
                               f"a reference is invalid, or a data type is wrong.",
                })

            # If no error in the evaluated result
            return json.dumps({
                "valid": True,
                "sample_result": format_value(sample_value),
                "evaluated_at": f"{sheet_name}!{target_cell}",
                "message": "Formula compiled and evaluated successfully.",
            })
        return json.dumps({ # if sample_value is None
            "valid": True,
            "message": "Formula compiled successfully (sample value not extracted).",
        })

    except Exception as e:
        return json.dumps({
            "valid": True,
            "message": f"Formula syntax is valid. Evaluation note: {e}",
        })
    finally:
        if os.path.exists(tmp_path):
            os.remove(tmp_path)
```

```python
def get_thread_history(agent, config: dict) -> list:
    """Peek at the messages stored in a thread (useful for debugging)."""
    try:
        state = agent.get_state(config)
        messages = state.values.get("messages", [])
        summary = []
        for msg in messages:
            entry = {"type": msg.type}
            if hasattr(msg, "tool_calls") and msg.tool_calls:
                entry["tool_calls"] = [tc["name"] for tc in msg.tool_calls]
            if msg.type == "tool":
                entry["content_preview"] = msg.content[:100]
            elif msg.type == "ai" and msg.content:
                entry["content_preview"] = msg.content[:100]
            elif msg.type == "human":
                entry["content_preview"] = msg.content[:100]
            summary.append(entry)
        return summary
    except Exception:
        return []

def print_thread_history(history: str) -> None:
    clean = [entry for entry in history if isinstance(entry, dict)]

    print(f"{'#':<4} {'Type':<8} {'Tool Calls':<35} {'Content Preview'}")
    print("-" * 120)
    for i, entry in enumerate(clean, 1):
        msg_type = entry.get("type", "")
        tools = ", ".join(entry.get("tool_calls", []))
        preview = entry.get("content_preview", "")[:60]
        print(f"{i:<4} {msg_type:<8} {tools:<35} {preview}")
```

```python
def analyze_thread(agent, config: dict) -> dict:
    """Analyze thread and return summary with token counts"""
    try:
        state = agent.get_state(config)
        messages = state.values.get("messages", [])

        total_messages = len(messages)
        total_tool_calls = 0
        total_input_tokens = 0
        total_output_tokens = 0

        message_details = []

        for msg in messages:
            msg_type = msg.type
            content = msg.content or ""

            content_tokens = len(content) // 4

            # Count calls
            tool_args_tokens = 0
            if msg_type == "ai" and hasattr(msg, "tool_calls") and msg.tool_calls:
                num_calls = len(msg.tool_calls)
                total_tool_calls += num_calls
                tool_names = [tc['name'] for tc in msg.tool_calls]
                for tc in msg.tool_calls:
                    args_str = str(tc.get("args", {}))
                    tool_args_tokens += len(args_str) // 4
            else:
                tool_names = []

            if msg_type in ["human", "tool"]:
                total_input_tokens += content_tokens
                direction = "input"
            elif msg_type == "ai":
                total_output_tokens += content_tokens + tool_args_tokens
                direction = "output"
            else:
                direction = "unknown"

            if tool_names:
                details = f"[calls: {', '.join(tool_names)}] content={content_tokens} + args={tool_args_tokens}"
            else:
                details = content[:80] + "..." if len(content) > 80 else content

            message_details.append({
                "type": msg_type,
                "direction": direction,
                "tokens": content_tokens + (tool_args_tokens if msg_type == "ai" else 0),
                "details": details
            })

        return {
            "total_messages": total_messages,
            "total_tool_calls": total_tool_calls,
            "total_input_tokens": total_input_tokens,
            "total_output_tokens": total_output_tokens,
            "total_tokens": total_input_tokens + total_output_tokens,
            "message_details": message_details
        }
    except Exception as e:
        return {"error": str(e)}

def print_summary(summary: dict) -> None:
    """Print summary in the format from your notebook"""
    if "error" in summary:
        print(f"Error: {summary['error']}")
        return

    print("="*70)
    print("THREAD ANALYSIS SUMMARY")
    print("="*70)
    print(f"  Total messages:       {summary['total_messages']}")
    print(f"  Total tool calls:     {summary['total_tool_calls']}")
    print(f"  Total input tokens:   {summary['total_input_tokens']}")
    print(f"  Total output tokens:  {summary['total_output_tokens']}")
    print(f"  Total tokens:         {summary['total_tokens']}")

    print("\n  PER-MESSAGE BREAKDOWN:")
    print(f"  {'#':<4} {'Role':<8} {'Direction':<10} {'Tokens':<8} Details")
    print("  " + "-"*70)

    for i, detail in enumerate(summary['message_details'], 1):
        print(f"  {i:<4} {detail['type']:<8} {detail['direction']:<10} {detail['tokens']:<8} {detail['details'][:60]}")

    print("="*70)

# Analysis
print("\n" + "="*80)
print("RUNNING THREAD ANALYSIS ON COMPLETE SESSION")
print("="*80)

summary = analyze_thread(excel_agent, excel_agent_config)
print_summary(summary)
```
